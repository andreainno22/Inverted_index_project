# PERFORMANCE ANALYSIS 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Standard Map-Reduce approach with different number of reducers

```python
# Create your dataset
data = [
    {"input_size": 100, "execution_time": 45.2, "reducers": 1},
    {"input_size": 100, "execution_time": 32.1, "reducers": 2},
    {"input_size": 100, "execution_time": 28.7, "reducers": 4},
    {"input_size": 200, "execution_time": 90.4, "reducers": 1},
    {"input_size": 200, "execution_time": 65.8, "reducers": 2},
    {"input_size": 200, "execution_time": 54.9, "reducers": 4},
    # add more entries as needed
]

# Convert to DataFrame
df = pd.DataFrame(data)

# Plot
plt.figure(figsize=(10, 6))
sns.lineplot(
    data=df,
    x="input_size",
    y="execution_time",
    hue="reducers",
    marker="o",
    palette="viridis"
)

plt.title("Input Size vs Execution Time (Colored by Reducers)")
plt.xlabel("Input Size (MB)")
plt.ylabel("Execution Time (s)")
plt.grid(True)
plt.legend(title="Reducers")
plt.tight_layout()
plt.show()
```
